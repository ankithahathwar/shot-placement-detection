```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    patience=20,        # stop early if no improvement for 20 epochs
    project="bullet_hole_detection",
    name="run1"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
import glob
sample_images = glob.glob(f"{dataset.location}/valid/images/*.jpg")[:3]
results = model.predict(sample_images, save=True, conf=0.4)
```

```python
import os

print(f"Contents of /content/:")
if os.path.exists('/content/'):
    print(os.listdir('/content/'))
else:
    print(f"Directory not found: /content/")

print(f"\nContents of /content/runs/:")
if os.path.exists('/content/runs/'):
    print(os.listdir('/content/runs/'))
else:
    print(f"Directory not found: /content/runs/")

# Define the base run directory where YOLO saves its runs
base_run_path = '/content/runs/detect/'

# List the contents of the base run directory to find the correct project folder
print(f"\nContents of {base_run_path}:")
if os.path.exists(base_run_path):
    project_folders = os.listdir(base_run_path)
    print(project_folders)

    # Try to find the 'bullet_hole_detection' folder
    if 'bullet_hole_detection' in project_folders:
        project_path = os.path.join(base_run_path, 'bullet_hole_detection')
        print(f"\nContents of {project_path}:")
        if os.path.exists(project_path):
            run_folders = os.listdir(project_path)
            print(run_folders)

            # Assuming the run folder is 'run1'
            run_path = os.path.join(project_path, 'run1')

            # If a 'weights' directory is found, list its contents as well
            weights_dir = os.path.join(run_path, 'weights')
            if os.path.exists(weights_dir):
                print(f"\nContents of {weights_dir}:")
                print(os.listdir(weights_dir))
            else:
                print(f"\nWeights directory not found at expected path: {weights_dir}")
                print(f"Listing contents of the run directory '{run_path}' to help locate weights:")
                if os.path.exists(run_path):
                    print(os.listdir(run_path))
                else:
                    print(f"Run directory not found: {run_path}")
        else:
            print(f"Project directory not found: {project_path}")
    else:
        print(f"'bullet_hole_detection' project folder not found in {base_run_path}")
else:
    print(f"Base run directory not found: {base_run_path}")
```
